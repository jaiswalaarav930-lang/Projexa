```python
import pandas as pd

# Load dataset
df = pd.read_csv("../dataset/student_study_data.csv")
```

```python
def generate_recommendation(row):
    # Calculate average marks
    avg_marks = (
        row['Math_Marks'] +
        row['Physics_Marks'] +
        row['Chemistry_Marks']
    ) / 3

    # Performance category
    if avg_marks >= 75:
        performance = "Good"
    elif avg_marks >= 50:
        performance = "Average"
    else:
        performance = "Poor"

    # Identify weak subjects
    weak_subjects = []
    if row['Math_Marks'] < 50:
        weak_subjects.append("Math")
    if row['Physics_Marks'] < 50:
        weak_subjects.append("Physics")
    if row['Chemistry_Marks'] < 50:
        weak_subjects.append("Chemistry")

    # Recommendations list
    recommendations = []

    if row['Attendance_Percentage'] < 75:
        recommendations.append("Improve class attendance")

    if row['Daily_Study_Hours'] < 2:
        recommendations.append("Increase daily study hours")

    if weak_subjects:
        recommendations.append(
            "Focus more on weak subjects: " + ", ".join(weak_subjects)
        )

    if performance == "Poor":
        recommendations.append("Revise basics and practice more questions")

    return performance, weak_subjects, recommendations
```

```python
# Example: Generate recommendation for a specific student
student_index = 1   # you can change this

performance, weak_subjects, recommendations = generate_recommendation(
    df.iloc[student_index]
)

print("Performance Category:", performance)
print("Weak Subjects:", weak_subjects)
print("Recommendations:")
for rec in recommendations:
    print("-", rec)
```

```python
df['Performance_Category'] = ""
df['Weak_Subjects'] = ""
df['Recommendations'] = ""

for i in range(len(df)):
    perf, weak, rec = generate_recommendation(df.iloc[i])
    df.at[i, 'Performance_Category'] = perf
    df.at[i, 'Weak_Subjects'] = ", ".join(weak)
    df.at[i, 'Recommendations'] = " | ".join(rec)

df.head()
```
